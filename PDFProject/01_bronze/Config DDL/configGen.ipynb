```sql
CREATE OR REPLACE TABLE ai.`02_bronze`.bronze_auditlog (
  id bigint GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
  loadid STRING,
  batchid STRING,
  sourcename STRING ,
  files_processed STRING ,
  status STRING,
  error_message STRING,
  loaddt TIMESTAMP,
  createdby STRING
  )
  CLUSTER BY (loaddt, loadid, batchid, sourceName)
```

```sql
CREATE TABLE IF NOT EXISTS ai.02_bronze.bronze_rejection_records (
    rejection_id        BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
    batch_id            BIGINT,
    rejected_payload    STRING,          -- Serialized JSON row data
    source_system       STRING,          -- Pipeline name, file path, or stream topic
    rejection_reason    STRING,          -- e.g., 'SCHEMA_MISMATCH', 'NULL_BUSINESS_KEY'
    rejected_at         TIMESTAMP,
    createdby STRING
)
USING DELTA
```
