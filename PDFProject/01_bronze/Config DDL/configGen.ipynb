```sql
CREATE OR REPLACE TABLE ai.`02_bronze`.bronze_auditlog (
  recordid BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
  loadid STRING COLLATE UTF8_BINARY,
  batchid STRING COLLATE UTF8_BINARY,
  sourcename STRING COLLATE UTF8_BINARY,
  files_processed STRING COLLATE UTF8_BINARY,
  valid_files STRING COLLATE UTF8_BINARY,
  invalid_files STRING COLLATE UTF8_BINARY,
  status STRING COLLATE UTF8_BINARY,
  error_message STRING COLLATE UTF8_BINARY,
  loaddt TIMESTAMP,
  createdby STRING COLLATE UTF8_BINARY)
USING delta
CLUSTER BY (loaddt,loadid,batchid,sourcename)
```

```sql
CREATE OR REPLACE TABLE ai.`02_bronze`.bronze_tutorial_document (
    recordid BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
  path STRING COLLATE UTF8_BINARY,
  modificationTime TIMESTAMP,
  length BIGINT,
  loaddt TIMESTAMP,
  createdby STRING COLLATE UTF8_BINARY,
  filename STRING COLLATE UTF8_BINARY,
  year INT,
  extension STRING COLLATE UTF8_BINARY,
  month INT,
  batchid INT,
  loadid STRING COLLATE UTF8_BINARY,
  documentKey STRING COLLATE UTF8_BINARY,
  isvalid BOOLEAN,
  validation_checks ARRAY<STRUCT<type: STRING COLLATE UTF8_BINARY, isvalid: BOOLEAN, error: STRING COLLATE UTF8_BINARY>>,
  mime_type STRING COLLATE UTF8_BINARY)
USING delta
CLUSTER BY (loadid, documentKey)
```
