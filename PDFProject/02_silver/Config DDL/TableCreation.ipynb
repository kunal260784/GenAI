```sql
CREATE OR REPLACE TABLE ai.`03_silver`.dim_document
(
    sourcefileid STRING, --source_arrival_id = f"{sourcename}:{loadid}:{bronze_record_id}"
    uuid STRING, --documentduplicatekey = str(    uuid.uuid5(ARRIVAL_NAMESPACE, source_arrival_id))
    sourcename STRING ,
    path STRING ,
    filename STRING,
    length BIGINT,
    extension STRING ,
    document_hash STRING ,
    mime_type STRING,
    page_count INT,
    createddt TIMESTAMP,
    createdby STRING ,
    modifieddt TIMESTAMP,
    modifiedby STRING ,
    loadid STRING
    )
USING DELTA
CLUSTER BY (sourcename,loadid,document_hash)
```
